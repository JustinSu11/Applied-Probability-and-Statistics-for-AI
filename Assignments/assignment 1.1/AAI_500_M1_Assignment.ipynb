Asked claude to help me with generating the synthetic rows

````python
print(gallstones['Gallstone Status'].value_counts())
numeric_cols = gallstones.select_dtypes(include='number').columns.drop(['Gallstone Status', 'Gender'])
synthetic = gallstones[positive].sample(50, replace=True, random_state=42).copy()
````

Add noise here

````python
for col in numeric_cols:
    noise = np.random.normal(0, gallstones[positive][col].std() * 0.05, size=50)
    synthetic[col] = synthetic[col].values + noise
augmented_gallstones = pd.concat([gallstones, synthetic], ignore_index=True)
print(augmented_gallstones['Gallstone Status'].value_counts())
````

````python
augmented_positive = augmented_gallstones['Gallstone Status'] == 1
augmented_negative = augmented_gallstones['Gallstone Status'] == 0
````

````python
plt.boxplot([augmented_gallstones[augmented_negative]['Body Mass Index (BMI)'], augmented_gallstones[augmented_positive]['Body Mass Index (BMI)']], vert=False, tick_labels=['Negative', 'Positive'])
plt.title('BMI by Gallstone Status')
plt.show()
````

````python
plt.boxplot([augmented_gallstones[augmented_negative]['Obesity (%)'], augmented_gallstones[augmented_positive]['Obesity (%)']], vert=False, tick_labels=['Negative', 'Positive'])
plt.title('Obesity by Gallstone Status')
plt.show()
````
